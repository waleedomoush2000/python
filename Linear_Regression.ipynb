```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path= '/content/drive/MyDrive/ML/students_scores.csv'
data=pd.read_csv(path)
```

```python
data
```

```python
x=data['Hours'].values
y=data['Scores'].values
print(x)
print(y)
print(x.shape)
print(y.shape)
```

```python
plt.figure(figsize=(10,5))
plt.scatter(x,y)
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.title('Hours vs Scores')
plt.show()
```

```python
L_rate=0.001
iteration=100
theata1=0
theata0=0
n=x.shape[0]
lossses=[]
for i in range(iteration):
  y_pred=theata1*x+theata0
  mse=(1/n)*np.sum((y_pred-y)**2)
  lossses.append(mse)
  dtheata1=(2/n)*np.sum(x*(y_pred-y))
  dtheata0=(2/n)*np.sum(y_pred-y)
  theata1=theata1-L_rate*dtheata1
  theata0=theata0-L_rate*dtheata0
```

```python
print('therat1 = ',theata1)
print('theta0 = ', theata0)
print('MSE = ',mse)
```

```python
new_x=9
predication_model=theata1*new_x+theata0
print('Hours = ',new_x)
print('Predicted Score = ',predication_model)
```

```python
plt.title('Lossses')
plt.plot(lossses)
plt.xlabel('Iteration')
plt.ylabel('Lossses')
plt.show()
print('initial loss', lossses[0])
print('final loss', lossses[-1] )
```

```python
x_line=np.linspace(0 ,10,100)
y_line=theata1*x_line+theata0
plt.figure(figsize=(10,5))
plt.scatter(x,y)
plt.plot(x_line,y_line,color='red')
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.title('Hours vs Scores')
```
